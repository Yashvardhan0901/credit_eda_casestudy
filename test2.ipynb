```python
import pandas as pd
```

```python
excel = pd.ExcelFile("Records Matching Task.xlsx")

print(excel.sheet_names)
```

```python
data1 = pd.read_excel("Records Matching Task.xlsx", sheet_name="data1")
data2 = pd.read_excel("Records Matching Task.xlsx", sheet_name="data2")
```

```python
print("Data1 Shape:", data1.shape)
print("Data2 Shape:", data2.shape)

data1.head()
```

```python
data2.head()
```

```python
data1["Key"] = (
    data1["Order ID"].astype(str)
    + "_"
    + data1["Product ID"].astype(str)
)

data2["Key"] = (
    data2["Order ID"].astype(str)
    + "_"
    + data2["Product ID"].astype(str)
)
```

```python
missing_in_data2 = data1[~data1["Key"].isin(data2["Key"])]

print("Records present in Data1 but missing in Data2:")
print(missing_in_data2.shape[0])
```

```python
missing_in_data1 = data2[~data2["Key"].isin(data1["Key"])]

print("Records present in Data2 but missing in Data1:")
print(missing_in_data1.shape[0])
```

```python
missing_in_data1["Qty"].sum()
```

```python
print("Sum of Qty:", missing_in_data1["Qty"].sum())
```

```python
combined = pd.concat([data1["Key"], data2["Key"]])

combined.nunique()
```
